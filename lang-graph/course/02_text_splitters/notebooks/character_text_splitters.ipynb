# Mastering Text Chunking Strategies with CharacterTextSplitter

In the realm of advanced Retrieval Augmented Generation (RAG), the quality of your retrieved context is paramount to generating accurate and coherent answers. However, large documents often exceed the token limits of LLMs or contain too much noise for effective retrieval. This is where text chunking becomes a critical engineering discipline. Simply splitting text by fixed character counts can lead to semantic breaks—cutting sentences or paragraphs mid-thought—which severely degrades the quality of the retrieved context.

This notebook dives deep into the `CharacterTextSplitter`, moving beyond basic splitting to explore sophisticated strategies like separator-based chunking, overlap management, and token-aware segmentation. Understanding how different separators (e.g., paragraph breaks vs. spaces) affect semantic integrity is crucial for building robust RAG pipelines. Furthermore, mastering the concept of chunk overlap allows us to ensure that critical information spanning a boundary is not lost during retrieval, significantly improving the completeness of the context provided to the LLM.

For advanced developers working with LangGraph and complex agents, effective chunking isn't just about size; it's about preserving *meaning*. By implementing token-aware splitters (like those based on tiktoken), we ensure that our chunks align with how the underlying language model processes text, leading to more predictable and higher-quality retrieval results. This knowledge is foundational for optimizing memory management and state transitions within complex graph structures.

### Text Splitter

### CharacterTextSplitter Initialization

This cell initializes a `CharacterTextSplitter`, which is responsible for breaking down large documents into smaller, manageable chunks. By setting `chunk_size=100` and `separator=""`, we ensure that each chunk will be exactly 100 characters long without any added separators.


```python
# create the splitter

# Initialize CharacterTextSplitter to break text into fixed-size chunks.
splitter = CharacterTextSplitter(
    chunk_size=100,  # Sets the maximum size of each resulting chunk (in characters).
    chunk_overlap=0, # Ensures no overlap between consecutive chunks.
    length_function=len, # Uses standard Python length calculation for chunk sizing.
    separator=""     # Specifies that no separator character should be used when splitting.
)

```

This cell executes the core text splitting process using the initialized `splitter` object. It calls `split_text(text)` to break the large document into smaller, manageable chunks (list of strings).
```python
# split the text

# Use the initialized splitter object to perform the actual splitting on the input text.
chunks = splitter.split_text(text)

# Print the resulting list of text chunks for inspection.
print(chunks)
# Print a summary count to verify that the chunking process was successful and quantify the output size.
print(f"Number of Chunks {len(chunks)}")
```

This function, `create_chunks`, demonstrates how to use a specific text splitter (`CharacterTextSplitter`) to break down large documents into smaller, manageable chunks. It is necessary for RAG systems because embedding models have token limits and chunking improves retrieval accuracy by focusing context.


```python
def create_chunks(text: str, chunk_size: int, 
                  separator: str, chunk_overlap: int = 0) -> list[str]:
    # Initialize the CharacterTextSplitter with specified parameters.
    splitter = CharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separator=separator
    )
    # Use the splitter's method to perform the actual text splitting.
    chunks = splitter.split_text(text=text)
    return chunks
```

This cell demonstrates the chunking process by calling `create_chunks` to split the original text into smaller segments (chunks) of 100 characters, and then passing these chunks to `display_chunks` for visualization. This step is crucial for preparing data for retrieval-augmented generation (RAG), as large documents must be broken down to fit within model context windows and improve search precision.


```python
display_chunks(create_chunks(text, 100, ""))
```

### Chunk Overlap

### Chunk Visualization Utility

This function visually demonstrates how chunk overlaps are handled in a list of text chunks. It is crucial for debugging and understanding the effectiveness of different text splitting strategies, as overlapping ensures context continuity between adjacent chunks.


```python
def display_chunks_with_overlaps(chunks, chunk_overlap):
    # Check if overlap size is zero, which would break the logic
    if not chunk_overlap:
        raise(ValueError("Chunk Overlap cannot be 0"))
    
    # Initialize variables for tracking and display
    overlap_text = ""
    overlap_color = "white"
    # Select a subset of colors for visualization (assuming COLORS is defined elsewhere)
    colors_list = list(COLORS.keys())[2:8]
    
    print(f"Total Number of Chunks: {len(chunks)}")
    # Iterate through the chunks, using enumerate to track the chunk number (starting at 1)
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        if num == 1:
            # For the first chunk, only overlap occurs at the end.
            # Print the non-overlapping start part of the chunk
            non_overlap_text = chunk[0:-chunk_overlap]
            # Extract the overlapping text from the end
            overlap_text = chunk[-chunk_overlap:]
            print(colored(non_overlap_text, choice(colors_list)),
                  colored(overlap_text,overlap_color),
                  sep="",end="\n\n")
        else:
            # For subsequent chunks, print the overlap text (which was saved from the previous chunk)
            print(colored(overlap_text, overlap_color), end="")
            # Extract and print the non-overlapping middle part of the current chunk
            non_overlap_text = chunk[chunk_overlap:-chunk_overlap]
            print(colored(non_overlap_text, choice(colors_list)), end="")
            # Update the overlap text for the next iteration (the last 'chunk_overlap' characters)
            overlap_text = chunk[-chunk_overlap:]
            # Print the newly determined overlap text at the end of the current chunk
            print(colored(overlap_text, overlap_color), end="\n\n")
```

### Token based splitting

### Token-Based Splitting using Tiktoken

This cell initializes a `CharacterTextSplitter` configured to split text based on token counts, specifically utilizing the `cl100k_base` encoding (used by OpenAI models). This ensures that chunks are sized according to how tokens are processed, which is crucial for maintaining context integrity and optimizing embedding costs.


```python
# create token based splitter

# Initialize CharacterTextSplitter using the tiktoken encoder.
# This method forces chunking based on token count rather than character count.
token_splitter = CharacterTextSplitter.from_tiktoken_encoder(
    encoding_name="cl100k_base", # Specifies the encoding used (e.g., OpenAI's standard)
    chunk_size=50,                 # Sets the target maximum size of each chunk in tokens
    chunk_overlap=5                # Defines the overlap between consecutive chunks to maintain context
)
```

```python
# split the text

token_splitter.split_text(text)
```
