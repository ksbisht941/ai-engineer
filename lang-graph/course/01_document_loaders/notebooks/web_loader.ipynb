```python
from langchain_community.document_loaders import WebBaseLoader
from pprint import pp
```

### Web Loader Initialization

This cell initializes a `WebBaseLoader` instance. This loader is designed to fetch content from multiple specified URLs (web paths) simultaneously, making it necessary for gathering diverse documentation or web data sources into a single processing unit.


```python
# create the loader

# Define the list of URLs/paths that will be loaded.
url_1 = "https://docs.langchain.com/oss/python/integrations/document_loaders/pypdfloader"
url_2 = "https://docs.langchain.com/oss/python/integrations/document_loaders/pdfminer"
url_3 = "https://docs.langchain.com/oss/python/integrations/document_loaders/pdfplumber"

# Initialize the WebBaseLoader, passing all defined URLs as web paths.
loader = WebBaseLoader(web_paths=[url_1, url_2, url_3])
```

### Document Loading

This cell executes the document loading process. The `loader` object, which was previously initialized (e.g., using a WebLoader), contains the logic to fetch and parse content from the source (like a webpage). Calling `.load()` triggers this process, populating the `documents` variable with a list of structured `Document` objects ready for embedding and retrieval.


```python
# load the document

# The loader object is called upon to execute its loading logic.
# This method fetches content from the source (e.g., a webpage).
# It returns a list of Document objects, which are stored in the 'documents' variable.
documents = loader.load()
```

### Code Explanation

```python
len(documents) # Calculates the length of the 'documents' list/variable, which represents the total count of loaded documents.
```

### Recursive Loader

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

### Web Content Loading

This cell initializes a `RecursiveUrlLoader` to fetch content from a specified URL (`base_url`). This loader is crucial for ingesting structured web documentation, recursively traversing up to two levels deep (`max_depth=2`) to gather comprehensive source material for the RAG system.


```python
# create the loader

# Define the base URL of the LangChain document loaders documentation.
base_url = "https://docs.langchain.com/oss/python/integrations/document_loaders"

# Initialize the RecursiveUrlLoader to fetch content from the specified URL.
# max_depth=2 ensures that the loader recursively crawls up to two levels deep, 
# gathering related documentation pages for comprehensive context.
loader = RecursiveUrlLoader(url=base_url, 
                            max_depth=2)
```

### Document Loading

This cell executes the document loading process. The `loader` object, which was previously configured (e.g., with specific paths or APIs), is called to perform its `load()` method, retrieving all necessary documents into the `documents` variable.


```python
# load the documents

documents = loader.load()
```

```python
# number of documents

len(documents)  # Calculates and prints the total count of document objects in the 'documents' list.
```

### Lazy Loading Documents

This step initializes a lazy loader, which is crucial for handling large datasets efficiently. Instead of loading all documents into memory immediately (which could cause memory overflow), `loader.lazy_load()` returns an object that only loads the document content when it is explicitly accessed or needed later in the pipeline.


```python
# lazy load

# Initialize the lazy loader object.
# This allows us to handle potentially massive datasets without loading all documents into memory at once,
# improving efficiency and preventing memory errors.
documents_lazy_load = loader.lazy_load()
```

### Code Explanation

This cell initializes the `documents_lazy_load` variable, which is expected to be a specialized loader or mechanism designed for efficient handling of large document sets. It likely implements lazy loading, meaning documents are only loaded into memory when they are actually needed, saving significant resources and time.


```python
documents_lazy_load
```

### Code Explanation

This cell iterates through a lazy-loaded collection of documents (`documents_lazy_load`), printing the beginning of each document's content and its associated metadata. It includes an explicit counter to ensure the loop processes exactly 20 documents, preventing potential memory issues or excessive runtime when dealing with large datasets.


```python
counter = 0

for document in documents_lazy_load:
    # stop condition: Break the loop once 20 documents have been processed.
    if counter == 20:
        break
    
    # increment the counter for tracking purposes.
    counter += 1
    
    # print metadata and page content (showing the first 300 characters of the document's content).
    print(document.page_content[0:300])
    pp(document.metadata)
```
