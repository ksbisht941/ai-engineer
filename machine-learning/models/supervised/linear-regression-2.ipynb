## Create Own Linear Regression Function

**Prediction (the line equation)**
$$\hat{y} = mx + b$$

**Slope**
$$m = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n}(x_i - \bar{x})^2}$$

**Intercept**
$$b = \bar{y} - m\bar{x}$$

### Example 2

```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
class LinearRegression:
    
    def __init__(self):
        self.weights = None   # Coefficients for all features
        self.bias = None      # Intercept term

    def fit(self, X_train, y_train):
        """
        X_train: shape (n_samples, n_features)
        y_train: shape (n_samples,)
        """

        # Convert to numpy arrays
        X_train = np.array(X_train)
        y_train = np.array(y_train)

        n_samples, n_features = X_train.shape

        # Normal Equation:
        # W = (X^T X)^(-1) X^T y

        # Add bias column of 1s
        X_b = np.c_[np.ones((n_samples, 1)), X_train]

        # Compute weights
        theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y_train)

        # Separate bias and weights
        self.bias = theta[0]
        self.weights = theta[1:]

        print("Weights:", self.weights)
        print("Bias:", self.bias)

    def predict(self, X_test):
        """
        X_test: shape (n_samples, n_features)
        """
        X_test = np.array(X_test)

        predictions = X_test.dot(self.weights) + self.bias

        return predictions
```

```python
class LinearRegressionSGD:
    
    def __init__(self, learning_rate=0.01, epochs=1000):
        # Step size for weight updates
        self.learning_rate = learning_rate
        
        # Number of full passes over data
        self.epochs = epochs
        
        # Model parameters
        self.weights = None
        self.bias = None

    def fit(self, X_train, y_train):
        
        # Convert input to NumPy arrays
        X_train = np.array(X_train)
        y_train = np.array(y_train)
        
        # Get number of samples and features
        n_samples, n_features = X_train.shape
        
        # Initialize weights as zeros
        self.weights = np.zeros(n_features)
        
        # Initialize bias
        self.bias = 0
        
        # Training loop
        for epoch in range(self.epochs):
            
            # Loop through each training sample
            for i in range(n_samples):
                
                # Select single sample
                x_i = X_train[i]
                y_i = y_train[i]
                
                # Prediction for current sample
                y_pred = np.dot(x_i, self.weights) + self.bias
                
                # Calculate gradients
                dw = -2 * x_i * (y_i - y_pred)
                db = -2 * (y_i - y_pred)
                
                # Update weights and bias
                self.weights = self.weights - self.learning_rate * dw
                self.bias = self.bias - self.learning_rate * db

            # Optional: print loss every 100 epochs
            if epoch % 100 == 0:
                predictions = np.dot(X_train, self.weights) + self.bias
                loss = np.mean((y_train - predictions) ** 2)
                print(f"Epoch {epoch}: Loss = {loss}")

    def predict(self, X_test):
        
        # Convert to NumPy array
        X_test = np.array(X_test)
        
        # Predict
        return np.dot(X_test, self.weights) + self.bias
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
lr = LinearRegressionSGD()
```

```python
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, y_pred)
```

```python
class LinearRegression:
    def __init__(self):
        self.intercept_ = None
        self.coef_ = None

    def fit(self, X, y):

        X = np.array(X)
        y = np.array(y)

        X_b = np.c_[np.ones((X.shape[0], 1)), X]

        theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

        self.intercept_ =  theta[0]
        self.coef_ =  theta[1:]

    def predict(self, X_test):
        X_test = np.array(X_test)
        return X_test.dot(self.coef_) + self.intercept_

lr = LinearRegression()
lr.fit(X_train, y_train)
lr.predict(X_test)
```
