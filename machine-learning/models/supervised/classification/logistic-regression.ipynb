```python
def step(z):
    # print("z", z)
    return 1 if z > 0 else 0
```

```python
def perceptron(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j], weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
    return weights[0],weights[1:]
```

```python
jdx = np.random.randint(0,100)
print(f"jdx: {jdx}")
weights = np.ones(X.shape[1])
print(f"weights: {weights}")
print(f"X[jdx]: {X[jdx]}")

print("="*20)

dot = np.dot(X[jdx], weights)
print(f"dot: {dot}")

print(f"step: {1 if dot > 0 else 0}")
```

```python
step(np.dot(X[jdx], weights))
```

```python
intercept_, coef_ = perceptron(X,y)

print("intercept_", intercept_)
print("coef_", coef_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```
