```python
from bs4 import BeautifulSoup
import requests
```

```python
headers = {'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
webpage = requests.get("https://www.ambitionbox.com/list-of-companies?page=1", headers=headers).text
```

```python
soup = BeautifulSoup(webpage, 'html.parser')
```

```python
# print(soup.prettify())
```

```python
soup.find_all('h1')[0].text
```

```python
company_headings = soup.find_all('h2', class_='companyCardWrapper__companyName')
for heading in company_headings:
    print(heading.get_text(strip=True))
```

```python
rating_text = soup.find_all('div', class_='rating_text')
for rating in rating_text:
    print(rating.get_text(strip=True))
```
